## Datos

El escenario siguiente es pequeño y explícito. `pointCount` es el número de puntos de órbita, indexados desde cero, y la partición cíclica incluye el cierre desde el punto mayor hasta el menor.

## Método

El cuaderno importa el motor del producto; no vuelve a implementar la asignación ni la agrupación de brechas.

The standard example uses the golden rotation, N=13, and phase 0.07.

El teorema de tres brechas supone una rotación irracional. Un parámetro finito en coma flotante no prueba irracionalidad; el certificado separa geometría finita y estado declarado de la entrada.

````python
# Import the canonical GapTheo engine from the repository pipeline.
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parents[2] / 'data-pipeline'))
from pipeline.gaptheo.model import certificate, allocate, gap_partition, continued_fraction, farey_bracket
scenario = {'id': 'golden-study', 'category': 'irrational', 'alpha': (5**0.5 - 1) / 2, 'alphaMode': 'irrational', 'pointCount': 13, 'phase': 0.07, 'allocator': 'rotation'}
result = certificate(scenario)
print(result['theoremStatus'], result['pointCount'], result['distinctCount'], result['partitionResidual'])
````

## Resultados

Lee las longitudes agrupadas y sus multiplicidades; luego verifica la suma cíclica completa y el estado de la entrada.

````python
# Compare a second view or a deliberately different regime using the same engine.
rational = dict(scenario, id='rational-boundary', alpha=5/13, alphaMode='rational', rationalP=5, rationalQ=13)
control = dict(scenario, id='farthest-control', allocator='farthest')
for case in (rational, control):
    cert = certificate(case)
    print(case['id'], cert['theoremStatus'], cert['distinctCount'])
````

## Resumen

La partición finita directa es reproducible, incluye la brecha de cierre y se distingue lógicamente de demostrar el teorema infinito.
